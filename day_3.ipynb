```python
import csv

with open("data_siswa.txt", "r") as f:
    reader = csv.reader(f)
    for baris in reader:
        nama = baris[0]
        nilai = int(baris[1])
        print(f"{nama} — nilai: {nilai} — {'Lulus' if nilai >= 70 else 'Tidak Lulus'}")
```
