```python
class Mahasiswa:
    def __init__(self, nama, nilai, jurusan):
        self.nama = nama
        self.nilai = nilai
        self.jurusan = jurusan

    def status(self):
        # sama persis seperti is_lulus() tadi
        # kalau nilai >= 70 return "Lulus", kalau tidak return "Tidak Lulus"
        if self.nilai >= 70:
            return "Lulus"
        else:
            return "Tidak Lulus"

    def perkenalan(self):
        # print "Halo, saya {nama} dari jurusan {jurusan}"
        return(f"Halo, saya {self.nama} dari jurusan {self.jurusan}")

# Buat object dari class
budi = Mahasiswa("Budi", 85, "Teknik INformatika")
ani = Mahasiswa("Ani", 60, "farmasi")

print(budi.nama)
print(budi.status())
print(ani.status())
print(budi.perkenalan())
print(ani.perkenalan())
```

```python
class MahasiswaBeasiswa(Mahasiswa):
    def __init__(self, nama, nilai, jurusan, nilai_beasiswa):
        super().__init__(nama, nilai, jurusan)
        self.nilai_beasiswa = nilai_beasiswa

andi = MahasiswaBeasiswa("Andi", 88, "Informatika", 5000000)

print(andi.perkenalan())   # method dari Mahasiswa
print(andi.status())       # method dari Mahasiswa
print(andi.nilai_beasiswa) # atribut baru
```

```python
class MahasiswaPrestasi(Mahasiswa):
    def __init__(self, nama, nilai, jurusan, ipk):
        super().__init__(nama, nilai, jurusan)
        self.ipk = ipk

    def predikat(self):
      if self.ipk >= 3.5 :
        return "CUmlaude"
      else :
        return "Reguler"

sandi = MahasiswaPrestasi("sandi", 90, "Informatika", 3.65)
print(sandi.predikat)    # ini akses atribut
print(sandi.predikat())  # ini panggil method — yang benar
```

```python
class Produk :
  def __init__(self, nama, harga, stok):
        self.nama = nama
        self.harga = harga
        self.stok = stok

  def tersedia(self):
    if self.stok > 0 :
      return True
    else:
      return False

  def beli(self, jumlah):
    if self.stok >= jumlah:
      self.stok -= jumlah
      print(f"Beli {jumlah} {self.nama} berhasil")
    else:
      print("Stock tidak cukup")


p = Produk("Sepatu", 150000, 5)
print(p.tersedia())  # True
p.beli(3)            # Beli 3 Sepatu berhasil
print(p.stok)        # 2
p.beli(10)           # Stock tidak cukup
```
